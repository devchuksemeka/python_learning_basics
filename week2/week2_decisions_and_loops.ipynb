## 1️⃣ Booleans: the computer's yes/no

A **boolean** is `True` or `False`. We get booleans by **comparing** things.

| Python | Meaning |
| ------ | ------- |
| `==` | equal to |
| `!=` | not equal to |
| `<` `>` | less / greater than |
| `<=` `>=` | less or equal / greater or equal |

⚠️ **Classic mistake:** `=` **puts** a value in a box. `==` **asks** "are these equal?"

````python
age = 15
print(age == 15)
print(age != 15)
print(age >= 18)
````

### Combining questions: `and`, `or`, `not`

**Analogy:** Entering a party. You need a ticket **AND** to be on the list. A guard says "VIPs **OR** staff can come in". **NOT** banned means everyone except banned people.

````python
has_ticket = True
on_list = False
print(has_ticket and on_list)   # both must be true
print(has_ticket or on_list)    # at least one
print(not on_list)              # flips it
````

## 2️⃣ Selection: `if`, `elif`, `else`

**Analogy:** A road junction. The program looks at a signpost and takes **one** road.

```python
if condition:
    do this
elif other_condition:
    do this instead
else:
    otherwise do this
```

- The line ends with a colon `:`
- The code inside is **indented** (4 spaces). Python uses spaces to know what belongs where.

````python
temperature = 34

if temperature >= 35:
    print("🔥 Too hot! Stay in the shade")
elif temperature >= 25:
    print("☀️ Nice weather")
else:
    print("🧥 Bring a jacket")
````

Python checks **top to bottom** and stops at the **first** true condition. Change `temperature` to 36, 20, 25 and predict the output *before* you run it.

## 3️⃣ Iteration: `for` loops

**Analogy:** "Do 10 push-ups." You know the count. 💪

`range(5)` gives `0, 1, 2, 3, 4` (starts at 0, stops *before* 5).

````python
for i in range(5):
    print("Push-up number", i)
````

````python
# range(start, stop, step)
for i in range(1, 11):
    print(i, end=" ")
print()
for i in range(0, 101, 10):
    print(i, end=" ")
````

### The totaling pattern (you'll use this everywhere)

> Start at **0** → **add** inside the loop → **print after** the loop

````python
total = 0
for i in range(1, 6):
    total = total + i
print("Total:", total)
````

````python
# 🎛️ Times-table maker
def times_table(n, upto):
    lines = []
    for i in range(1, int(upto) + 1):
        lines.append(f"{int(n)} x {i} = {int(n) * i}")
    return "\n".join(lines)

demo = gr.Interface(
    fn=times_table,
    inputs=[
        gr.Number(label="Which table?", value=7),
        gr.Slider(1, 20, value=12, step=1, label="Up to"),
    ],
    outputs=gr.Textbox(label="Your times table", lines=12),
    title="✖️ Times-Table Maker",
)
show(demo)
````

## 4️⃣ Iteration: `while` loops

Use `while` when you **don't know** how many times. It repeats **until a condition becomes false**.

**Analogy:** "Keep stirring until the sauce is thick." 🥣

⚠️ **Danger:** if the condition never becomes false you get an **infinite loop**. (In Jupyter, press the ■ stop button to escape.)

````python
def weeks_to_goal(goal, weekly):
    if weekly <= 0:
        return "Weekly saving must be more than 0!"
    saved = 0
    weeks = 0
    while saved < goal:
        saved = saved + weekly
        weeks = weeks + 1
    return f"You will reach ₦{goal:,.0f} in {weeks} weeks 🎯"

demo = gr.Interface(
    fn=weeks_to_goal,
    inputs=[
        gr.Number(label="Savings goal (₦)", value=20000),
        gr.Number(label="Save per week (₦)", value=1500),
    ],
    outputs="text",
    title="🎯 Goal Tracker",
)
show(demo)
````

### `for` vs `while`

| `for` | `while` |
| ----- | ------- |
| Known number of repeats | Unknown number of repeats |
| Python counts for you | You manage the condition |
| "Do this 10 times" | "Keep going until …" |

### Finished version (reveal after class)

````python
def check_password(password):
    has_digit = False
    has_capital = False
    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch.isupper():
            has_capital = True

    score = 0
    tips = []
    if len(password) >= 8:
        score += 1
    else:
        tips.append("Use at least 8 characters")
    if has_digit:
        score += 1
    else:
        tips.append("Add a number")
    if has_capital:
        score += 1
    else:
        tips.append("Add a capital letter")

    if score == 3:
        return "💪 Strong password!"
    return f"Score {score}/3. " + " · ".join(tips)

demo = gr.Interface(
    fn=check_password,
    inputs=gr.Textbox(label="Type a test password (not your real one!)"),
    outputs="text",
    title="🔐 Password Checker",
    live=True,
)
show(demo)
````
