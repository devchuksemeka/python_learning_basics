## 1️⃣ How the web works: the restaurant 🍽️

| In the restaurant | On the web |
| ----------------- | ---------- |
| You (the customer) | Your **browser** (Chrome, Safari...) |
| The waiter | The **internet** |
| The kitchen | The **server** |
| The menu you read | **HTML** (structure and content) |
| How the restaurant looks (decor, colours) | **CSS** (style) |
| The magic tricks (flaming desserts!) | **JavaScript** (behaviour, Week 8) |

When you type `https://academy.carojon.com/lessons?id=7`:
- `https` → the **protocol** (secure way to talk)
- `academy.carojon.com` → the **domain** (which server)
- `/lessons` → the **path** (which page)
- `?id=7` → the **query** (extra options, like parameters!)

**Remember Gradio?** When you ran `show(demo)`, Gradio started a tiny **server** inside your computer, and your notebook displayed the **web page it generated** (HTML + CSS + JavaScript). Today you write those layers **yourself**.

## 2️⃣ HTML — the skeleton 🦴

**HTML** = **H**yper**T**ext **M**arkup **L**anguage. It's **not** a programming language, it **describes** a page using **tags**.

```html
<tagname attribute="value">content</tagname>
```

- Most tags **open** (`<p>`) and **close** (`</p>`)
- Tags can be **nested** (a list *inside* the body)
- **Attributes** give extra info (`href`, `src`, `alt`)

### 🛠️ How we run HTML in this notebook
Start a cell with **`%%web`** and write HTML. The page appears underneath. (`%%web 400` = 400px tall)

````python
%%web
<h1>Hello, web! 🌍</h1>
<p>This is my first <strong>real</strong> web page.</p>
````

### The most useful tags

| Tag | What it makes |
| --- | ------------- |
| `<h1>` … `<h6>` | headings (h1 biggest) |
| `<p>` | paragraph |
| `<strong>` / `<em>` | **bold** / *italic* |
| `<ul>` + `<li>` | bullet list |
| `<ol>` + `<li>` | numbered list |
| `<a href="...">` | link |
| `<img src="..." alt="...">` | image |
| `<button>` | button |
| `<br>` / `<hr>` | line break / line |

````python
%%web 330
<h1>About Me</h1>
<p>My name is <strong>Ada</strong> and I love <em>football</em>.</p>

<h2>My hobbies</h2>
<ul>
  <li>Football</li>
  <li>Drawing</li>
  <li>Coding</li>
</ul>

<h2>My top 3 foods</h2>
<ol>
  <li>Jollof rice</li>
  <li>Suya</li>
  <li>Moi moi</li>
</ol>

<a href="https://academy.carojon.com">Visit Carojon Academy</a>
````

### The page skeleton

```html
<!DOCTYPE html>
<html>
  <head>   <!-- BACKSTAGE: title, settings -->
    <title>My Page</title>
  </head>
  <body>   <!-- THE STAGE: what visitors see -->
    ...
  </body>
</html>
```

**Analogy:** A theatre. The **head** is backstage (the audience doesn't see it). The **body** is the stage.

### Forms: how pages collect information 📝
Remember Gradio's text boxes and buttons? Here they are in raw HTML.

````python
%%web 300
<h2>Join the club</h2>
<form>
  <label>Your name:</label><br>
  <input type="text" placeholder="Type your name"><br><br>

  <label>Favourite subject:</label><br>
  <select>
    <option>Maths</option>
    <option>Science</option>
    <option>Coding</option>
  </select><br><br>

  <label>Age: <input type="number" min="5" max="20" value="12"></label><br><br>
  <button type="button">Join!</button>
</form>
````

### ♿ Accessibility: a website for *everyone*
- Always add **`alt`** text to images (for people using screen readers)
- Use **headings in order** (h1 → h2 → h3)
- Give inputs a **`<label>`**

*(Remember the digital divide from the exam course: good developers build for everyone.)*

## 3️⃣ CSS — the clothes and decoration 🎨

**CSS** = **C**ascading **S**tyle **S**heets. HTML says **what** things are, CSS says **how they look**.

```css
selector {
  property: value;
}
```

**Analogy:** HTML is a person. CSS is their **clothes, hairstyle and accessories**. 👗

````python
%%web 280
<style>
  h1 {
    color: darkgreen;
    font-family: Arial, sans-serif;
    text-align: center;
  }
  p {
    font-size: 20px;
    color: #444444;
  }
  .highlight {            /* a CLASS: starts with a dot */
    background: yellow;
    padding: 4px;
  }
</style>

<h1>Styled Page</h1>
<p>Normal text.</p>
<p class="highlight">This paragraph is highlighted!</p>
````

### Selectors: who gets styled?

| Selector | Targets | Example |
| -------- | ------- | ------- |
| `p` | every `<p>` tag | `p { color: blue; }` |
| `.name` | everything with `class="name"` | `.card { ... }` |
| `#name` | the one thing with `id="name"` | `#title { ... }` |

### Colours: remember hex? 🎨
`#FF5733` is **three hex numbers**: Red `FF`, Green `57`, Blue `33` (each 0–255). **The same hex you learned about computers!**

````python
# Python can convert colours too
r, g, b = 255, 87, 51
print(f"#{r:02X}{g:02X}{b:02X}")
print(int("FF", 16), int("57", 16), int("33", 16))
````

## 4️⃣ The Box Model 📦

Every HTML element is a **box**.

**Analogy:** A gift parcel. 🎁
- **Content**: the gift
- **Padding**: the bubble-wrap (space *inside*, around content)
- **Border**: the box wall
- **Margin**: space *outside* between this parcel and other parcels

````python
%%web 330
<style>
  .card {
    background: #e8f5e9;
    border: 3px solid #2e7d32;
    border-radius: 16px;       /* rounded corners */
    padding: 20px;             /* inside space */
    margin: 20px;              /* outside space */
    width: 250px;
    font-family: Arial;
  }
</style>

<div class="card">
  <h3>Ada's Card</h3>
  <p>Change padding, margin, border-radius and see what moves!</p>
</div>
````

## 5️⃣ Layout with Flexbox 🧩

`display: flex` puts items in a **row** and lets you control spacing.

**Analogy:** Friends sitting on a bench. 🪑 You can push them to the left, spread them out, or centre them.

````python
%%web 280
<style>
  .row {
    display: flex;
    gap: 16px;                       /* space between friends */
    justify-content: center;         /* try: flex-start, space-between, space-around */
    flex-wrap: wrap;                 /* wrap on small screens */
    font-family: Arial;
  }
  .box {
    background: #1976d2;
    color: white;
    padding: 24px;
    border-radius: 12px;
    width: 120px;
    text-align: center;
    transition: transform 0.2s;
  }
  .box:hover { transform: scale(1.1); background: #0d47a1; }   /* hover effect! */
</style>

<div class="row">
  <div class="box">🍚 Rice</div>
  <div class="box">🍗 Chicken</div>
  <div class="box">🥤 Drink</div>
</div>
````

### 📱 Responsive design
Your page should look good on a **phone, tablet and laptop**. A **media query** changes styles by screen width.

````python
%%web 300
<style>
  body { font-family: Arial; }
  .banner { padding: 20px; background: #6a1b9a; color: white; text-align: center; font-size: 32px; }

  /* On screens narrower than 500px, make the text smaller and change colour */
  @media (max-width: 500px) {
    .banner { font-size: 18px; background: #ad1457; }
  }
</style>
<div class="banner">I change on small screens!</div>
<p>Drag the notebook sidebar wider/narrower or open the saved page in your browser and resize the window.</p>
````

## 6️⃣ 🐍 + 🌐: Python can WRITE web pages

Remember f-strings and loops? Python can **generate HTML**. This is how many real websites are built.

````python
from IPython.display import HTML

students = [
    {"name": "Ada", "hobby": "Football"},
    {"name": "Tobi", "hobby": "Drawing"},
    {"name": "Chi", "hobby": "Coding"},
]

cards = ""
for s in students:
    cards += f"""
    <div style="background:#fff3e0;border:2px solid #ef6c00;border-radius:12px;padding:14px;width:140px;text-align:center;font-family:Arial">
        <h3>{s['name']}</h3><p>Loves {s['hobby']}</p>
    </div>"""

page = f"<div style='display:flex;gap:12px;flex-wrap:wrap'>{cards}</div>"
HTML(page)
````

````python
# The same thing inside a Gradio app!
def make_cards(names_text):
    names = [n.strip() for n in names_text.split(",") if n.strip() != ""]
    out = ""
    for n in names:
        out += f"<div style='display:inline-block;margin:6px;padding:14px 20px;background:#e3f2fd;border-radius:12px;font-family:Arial'>👤 {n}</div>"
    return out

demo = gr.Interface(
    fn=make_cards,
    inputs=gr.Textbox(label="Names, separated by commas", value="Ada, Tobi, Chi"),
    outputs=gr.HTML(),
    title="🪪 Name Card Maker",
    flagging_mode="never",
)
show(demo)
````
