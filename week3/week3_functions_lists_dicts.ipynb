## 1️⃣ Functions — build your own tools

You've already *used* functions: `print()`, `type()`, `range()`. Now you **make** your own.

**Analogy:** A kitchen blender.
- Put **ingredients** in → these are the **parameters**
- Press the button → **calling** the function
- Get a **smoothie** out → the **return value**

```python
def blend(fruit1, fruit2):
    return f"{fruit1}-{fruit2} smoothie 🥤"
```

````python
def blend(fruit1, fruit2):
    return f"{fruit1}-{fruit2} smoothie 🥤"

print(blend("mango", "banana"))
print(blend("pineapple", "orange"))
````

### Parameters vs arguments vs return

| Word | Meaning | In the example |
| ---- | ------- | -------------- |
| **Parameter** | the name in the definition | `fruit1` |
| **Argument** | the real value you pass in | `"mango"` |
| **Return value** | what the function hands back | the smoothie text |

⚠️ **`print` vs `return`:** `print` *shows* something on screen. `return` *hands a value back* so other code can use it. **Gradio apps need `return`.**

````python
def area(length, width):
    return length * width

room = area(4, 5)
print(room)
print(area(2, 3) + area(1, 1))   # we can USE the returned value
````

````python
# Default values: if you don't pass an argument, the default is used
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Ada"))
print(greet("Ada", "Good morning"))
````

### Why functions?
- **Reuse:** write once, use 100 times
- **Easier to read and fix**
- **Teamwork:** different people write different functions
- **Every Gradio app is a function!**

### Local vs global variables
A variable made **inside** a function lives only inside it (like notes in your own exercise book). A variable **outside** is visible to everyone (like the school notice board). Prefer **local** variables; pass things in with parameters.

## 2️⃣ Lists — one box, many items

**Analogy:** A shopping list. 🛒 Items are in order, and each has a position number called an **index**. **Counting starts at 0!**

````python
fruits = ["apple", "banana", "mango"]

print(fruits[0])     # first item
print(fruits[2])     # third item
print(fruits[-1])    # last item
print(len(fruits))   # how many
````

````python
fruits.append("pawpaw")      # add to the end
fruits[1] = "pear"           # change an item
fruits.remove("apple")       # remove by value
print(fruits)
print("mango" in fruits)     # is it in the list?
````

````python
# Loop through a list: the most useful pattern in Python
scores = [45, 78, 62, 90]

total = 0
for s in scores:
    total = total + s

print("Total:", total)
print("Average:", total / len(scores))
print("Highest:", max(scores), "| Lowest:", min(scores))
````

### 🎛️ Live-tweak: Business-Name Generator
Lists + `random.choice` = instant creativity. **Tutor: edit the lists live** (add local words, slang, food, anything).

````python
adjectives = ["Golden", "Royal", "Glow", "Smooth", "Fresh", "Bright", "Classy"]
nouns = ["Touch", "Studio", "Corner", "Palace", "Hub", "Lounge", "Spot"]

def business_name(kind):
    adj = random.choice(adjectives)
    noun = random.choice(nouns)
    return f"{adj} {noun} {kind}"

demo = gr.Interface(
    fn=business_name,
    inputs=gr.Dropdown(["Salon", "Bakery", "Tailors", "Barbers", "Cafe"], label="Type of business", value="Salon"),
    outputs=gr.Textbox(label="Your business name"),
    title="🏷️ Business Name Generator",
    flagging_mode="never",
)
show(demo)
````

## 3️⃣ Dictionaries — look things up by name

**Analogy:** A real dictionary or a phone book. You look up the **key** (the word) and get the **value** (the meaning).

Lists use **numbers** to find things. Dictionaries use **names**.

````python
student = {"name": "Ada", "age": 12, "school": "Carojon Academy"}

print(student["name"])
student["age"] = 13               # update
student["favourite_game"] = "chess"   # add a new key
print(student)
````

````python
for key, value in student.items():
    print(key, "→", value)
````

### The power move: a **list of dictionaries** = a table

| name | age | score |
| ---- | --- | ----- |
| Ada | 12 | 88 |
| Tobi | 15 | 72 |
| Chi | 13 | 95 |

This is how almost all real data is stored (and how AI services send you data).

````python
students = [
    {"name": "Ada", "age": 12, "score": 88},
    {"name": "Tobi", "age": 15, "score": 72},
    {"name": "Chi", "age": 13, "score": 95},
]

best = students[0]
for s in students:
    if s["score"] > best["score"]:
        best = s
print("Top student:", best["name"], "with", best["score"])
````

````python
def lookup(name):
    for s in students:
        if s["name"] == name:
            return f"{s['name']} is {s['age']} years old and scored {s['score']}%"
    return "Not found 🤷"

demo = gr.Interface(
    fn=lookup,
    inputs=gr.Dropdown([s["name"] for s in students], label="Pick a student"),
    outputs="text",
    title="📒 Class Register",
    flagging_mode="never",
)
show(demo)
````

## 4️⃣ Mini project: Flashcards (functions + dictionary + `if`)
A dictionary of questions and answers is the **brain**; Gradio is the **face**.

This one uses **Blocks**: a way to place components (text boxes, buttons) wherever we like.

````python
cards = {
    "What does print() do?": "shows something",
    "What is 2 ** 3?": "8",
    "What type is 3.14?": "float",
    "What does len([4, 5, 6]) give?": "3",
}

def new_question():
    return random.choice(list(cards)), "", ""

def check(question, answer):
    if question == "":
        return "Press 'New question' first 🙂"
    right = cards[question]
    if answer.strip().lower() == right.lower():
        return "✅ Correct!"
    return f"❌ Not quite. The answer was: {right}"

with gr.Blocks(title="Flashcards") as demo:
    gr.Markdown("# 🃏 Python Flashcards")
    question = gr.Textbox(label="Question", interactive=False)
    answer = gr.Textbox(label="Your answer")
    result = gr.Markdown()
    with gr.Row():
        gr.Button("New question").click(new_question, None, [question, answer, result])
        gr.Button("Check answer", variant="primary").click(check, [question, answer], result)

show(demo)
````
