## 2️⃣ Variables — labelled boxes

A **variable** is a box with a name on it. You put a value inside, and you can use it later.

**Analogy:** A school locker. The locker has a **name** (`locker_12`) and **something inside** it. You can swap what's inside.

````python
name = "Ada"
age = 12
favourite_game = "football"

print(name)
print(age)
print("My favourite game is", favourite_game)
````

````python
# Change what's inside the box
age = 13
print(name, "is now", age)
````

### Naming rules (very important!)
- Use **meaningful** names: `pocket_money`, not `x`
- Use `snake_case` (words joined with `_`)
- No spaces, can't start with a number
- `Name` and `name` are **different** boxes

## 3️⃣ Data types — different kinds of things

| Type | Python word | What it stores | Example |
| ---- | ----------- | -------------- | ------- |
| Text | `str` (string) | words, sentences | `"Hello"` |
| Whole number | `int` | counting numbers | `42` |
| Decimal number | `float` | numbers with a point | `3.14` |
| True / False | `bool` | yes or no | `True` |

**Analogy:** Containers. A glass holds water, a box holds books. Each type holds a different kind of thing.

````python
print(type("Hello"))
print(type(42))
print(type(3.14))
print(type(True))
````

````python
# Text + text = joined text. Number + number = maths!
print("5" + "5")
print(5 + 5)
````

👀 Notice: `"5" + "5"` gave `55` (joined) but `5 + 5` gave `10` (added). **The type decides what an operator does.**

## 4️⃣ Python as a super calculator

| Operator | Meaning | Example | Result |
| -------- | ------- | ------- | ------ |
| `+` | add | `7 + 2` | 9 |
| `-` | subtract | `7 - 2` | 5 |
| `*` | multiply | `7 * 2` | 14 |
| `/` | divide | `7 / 2` | 3.5 |
| `//` | divide, whole number only | `7 // 2` | 3 |
| `%` | **remainder** | `7 % 2` | 1 |
| `**` | power | `7 ** 2` | 49 |

**Analogy for `//` and `%`:** Share 7 sweets between 2 friends. Each gets **3** (`//`) and **1** is left over (`%`).

````python
print(7 // 2)
print(7 % 2)
print(2 ** 10)
````

## 5️⃣ 🎉 Big idea: from code to a real app

Every app you use does three things:

```
   INPUT   ───►   FUNCTION   ───►   OUTPUT
 (what you      (the thinking       (what you
  give it)       part - Python)      get back)
```

**Analogy:** A juice blender. Put **fruit** in (input), the **blender** does its work (function), you get **juice** out (output).

- A **function** is a named block of code. We will study functions properly in Week 3. Today: just **spot the pattern**.
- **Gradio** is a tool that wraps your function in a nice screen, so *anyone* can use it.

````python
import gradio as gr

def greet(name):
    return f"Hello, {name}! Welcome to Carojon Academy 🎉"

demo = gr.Interface(
    fn=greet,
    inputs="text",
    outputs="text",
    title="👋 My First App",
)
demo.launch()
````

### 🔍 Anatomy of the app
- `def greet(name):` → we **define** a function called `greet`. `name` is what goes **in**.
- `return ...` → what comes **out**.
- `gr.Interface(fn=..., inputs=..., outputs=...)` → builds the screen.
- `show(demo)` → shows it right here in the notebook.

### 🎛️ Live-tweak time (change ONE thing, rerun the cell, see what happens)
1. Change the message to something funny.
2. Change `title=` to your own app name.
3. Make it shout: replace `name` with `name.upper()`.

````python
# Two inputs! Notice how the inputs list matches the function's parameters, in order.
def future_age(name, age):
    return f"{name}, in 10 years you will be {age + 10}! 🚀"

demo = gr.Interface(
    fn=future_age,
    inputs=[gr.Textbox(label="Your name"), gr.Number(label="Your age", value=12)],
    outputs=gr.Textbox(label="Your future"),
    title="🔮 Age Machine",
)
show(demo)
````

````python
# Sliders and money 💰
def savings(weekly_amount, weeks):
    total = weekly_amount * weeks
    return f"In {int(weeks)} weeks you will have ₦{total:,.0f} 💰"

demo = gr.Interface(
    fn=savings,
    inputs=[
        gr.Number(label="Weekly pocket money (₦)", value=500),
        gr.Slider(1, 52, value=10, step=1, label="Number of weeks"),
    ],
    outputs=gr.Textbox(label="Total saved"),
    title="💰 Savings Calculator",
)
show(demo)
````
