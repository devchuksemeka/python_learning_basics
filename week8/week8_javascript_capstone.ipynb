## 1️⃣ JavaScript: the muscles and brain 🧠

| Layer | Job | Analogy |
| ----- | --- | ------- |
| **HTML** | Structure | Skeleton 🦴 |
| **CSS** | Style | Clothes 👗 |
| **JavaScript** | Behaviour | Muscles and brain 💪🧠 |

JavaScript (**JS**) runs **inside the browser**. It makes buttons *do things*, updates text without reloading, plays games, and more.

**Good news:** you already know programming. JS has the same ideas as Python, just different spelling!

| Idea | Python | JavaScript |
| ---- | ------ | ---------- |
| Variable | `age = 12` | `let age = 12;` |
| Constant | `PI = 3.14` | `const PI = 3.14;` |
| Print | `print("Hi")` | `console.log("Hi");` |
| If | `if x > 5:` | `if (x > 5) {` |
| Else if | `elif` | `else if` |
| For loop | `for i in range(5):` | `for (let i = 0; i < 5; i++) {` |
| Function | `def add(a, b):` | `function add(a, b) {` |
| List | `[1, 2, 3]` | `[1, 2, 3]` |
| Dictionary | `{"name": "Ada"}` | `{ name: "Ada" }` |
| Joining text | `f"Hi {name}"` | `` `Hi ${name}` `` |

Notice: JS uses **curly braces `{ }`** and **semicolons `;`** instead of indentation and colons.

````python
%%console
function grade(mark) {
  if (mark >= 70) {
    return "A";
  } else if (mark >= 50) {
    return "C";
  } else {
    return "Keep going";
  }
}

console.log(grade(85));
console.log(grade(40));

const scores = [45, 78, 62, 90];
const passed = scores.filter(s => s >= 50);     // keep only some items
const doubled = scores.map(s => s * 2);         // change every item
console.log(passed);
console.log(doubled);
````

## 2️⃣ The DOM: JavaScript's remote control 🎮

The browser turns your HTML into a **tree of objects** called the **DOM** (**D**ocument **O**bject **M**odel). JavaScript can **find** any piece and **change** it.

**Analogy:** A TV remote. The TV is your page. 📺 The remote (JS) can change channels, volume and brightness.

**Events** are things that happen: a click, a key press, a typing. We tell JS: **"When THIS happens, do THAT."**

**Analogy:** A **doorbell**. 🔔 When it rings (event), you go and open the door (the function).

````python
%%web 220
<h2 id="count">0</h2>
<button id="btn">Click me!</button>

<script>
  let count = 0;
  const display = document.getElementById("count");      // FIND the element

  document.getElementById("btn").addEventListener("click", function () {   // WHEN clicked...
    count = count + 1;
    display.textContent = count;                          // ...CHANGE it
  });
</script>
````

````python
%%web 300
<style>
  body { font-family: Arial; transition: 0.3s; padding: 16px; }
  .dark { background: #212121; color: white; }
</style>

<h2>🌙 Dark mode switch</h2>
<p>Click the button to toggle the theme.</p>
<button id="toggle">Toggle theme</button>

<script>
  document.getElementById("toggle").addEventListener("click", () => {
    document.body.classList.toggle("dark");
  });
</script>
````

````python
%%web 340
<style>
  body { font-family: Arial; }
  li { margin: 6px 0; }
</style>

<h2>✅ To-do list</h2>
<input id="task" placeholder="New task...">
<button id="add">Add</button>
<ul id="list"></ul>

<script>
  const input = document.getElementById("task");
  const list = document.getElementById("list");

  document.getElementById("add").addEventListener("click", () => {
    if (input.value.trim() === "") {
      return;                                    // validation, just like Python!
    }
    const item = document.createElement("li");   // make a NEW element
    item.textContent = input.value;
    list.appendChild(item);                      // put it on the page
    input.value = "";
  });
</script>
````

## 3️⃣ JS can call APIs too 🌦️

Week 5's idea, now **in the browser**: `fetch` is JavaScript's version of `requests.get`. Because the network takes time, we use **`async/await`**:

**Analogy:** Order food 🍔 → **wait** for it (`await`) → then eat.

````python
%%web 220
<h3>Weather in Awka 🌤️</h3>
<p id="out">Loading...</p>

<script>
  async function getWeather() {
    try {
      const url = "https://api.open-meteo.com/v1/forecast?latitude=6.21&longitude=7.07&current_weather=true";
      const response = await fetch(url);
      const data = await response.json();            // same JSON you know!
      document.getElementById("out").textContent =
        "Temperature now: " + data.current_weather.temperature + " °C";
    } catch (error) {
      document.getElementById("out").textContent = "Could not reach the weather service 😢";
    }
  }
  getWeather();
</script>
````

### 🔐 Important security rule
If a web page calls an AI API **directly from JavaScript**, your **secret key** would be visible to *anyone* (they can right-click → "View source"). So real apps do this:

```
Browser (HTML/CSS/JS)  ──►  YOUR server (Python, hides the key)  ──►  AI API
```

**This is exactly what Gradio did for you in Week 6.** Python ran on the server side with the key safely hidden, and Gradio showed the page.
