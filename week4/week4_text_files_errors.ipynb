## 1️⃣ Strings — text is a list of characters

**Analogy:** A string is a **necklace of beads**. 📿 Each bead (character) has a position, starting at 0.

```python
word = "Programming"

print(len(word))       # how many characters
print(word[0])         # first bead
print(word[-1])        # last bead
print(word[0:4])       # slice: beads 0,1,2,3  → "Prog"
print(word[::-1])      # a slice with step -1 → reversed!
```

```python
sentence = "  Python is Fun  "

print(sentence.upper())
print(sentence.lower())
print(sentence.strip())             # removes spaces at both ends
print(sentence.replace("Fun", "AWESOME"))
print("Python" in sentence)         # is it inside?
print(sentence.strip().split(" "))  # cut into a list of words
```

```python
words = ["red", "green", "blue"]
print(" - ".join(words))            # join is the opposite of split
```

**Strings are immutable** (can't be changed in place), like a printed certificate. Methods like `.upper()` give you a **new** string.

| Method | What it does |
| ------ | ------------ |
| `.upper()` / `.lower()` | change case |
| `.strip()` | remove spaces at the ends |
| `.split(sep)` | text → list |
| `"sep".join(list)` | list → text |
| `.replace(a, b)` | swap text |
| `.find(x)` | position of x (or -1) |
| `.isdigit()` / `.isalpha()` | what kind of characters |

```python
# 🎛️ Live-tweak: Text Toolbox with tabs
def count_words(text):
    words = text.split()
    return f"{len(words)} words · {len(text)} characters"

def is_palindrome(text):
    cleaned = text.lower().replace(" ", "")
    if cleaned == cleaned[::-1]:
        return f"✅ '{text}' reads the same backwards!"
    return f"❌ '{text}' is not a palindrome"

def caesar(text, shift):
    result = ""
    for ch in text:
        if ch.isalpha():
            base = ord("A") if ch.isupper() else ord("a")
            result += chr((ord(ch) - base + int(shift)) % 26 + base)
        else:
            result += ch
    return result

with gr.Blocks(title="Text Toolbox") as demo:
    gr.Markdown("# 🧰 Text Toolbox")
    with gr.Tab("Word counter"):
        t1 = gr.Textbox(lines=4, label="Paste some text")
        o1 = gr.Textbox(label="Result")
        t1.change(count_words, t1, o1)
    with gr.Tab("Palindrome checker"):
        t2 = gr.Textbox(label="Try: Madam, level, or racecar")
        o2 = gr.Textbox(label="Result")
        t2.change(is_palindrome, t2, o2)
    with gr.Tab("Secret code 🕵️"):
        t3 = gr.Textbox(label="Message")
        s3 = gr.Slider(1, 25, value=3, step=1, label="Shift (the secret key)")
        o3 = gr.Textbox(label="Coded message")
        gr.Button("Encode").click(caesar, [t3, s3], o3)

show(demo, height=420)
```

### 🕵️ How the secret code works
`ord("a")` gives the **number** the computer uses for the letter `a` (97). `chr(97)` goes back. The Caesar cipher shifts each letter along the alphabet, and `% 26` makes it **wrap around** (remember `%`?). To decode, shift by `26 - shift`.

**Link to Week 1:** every letter is stored as a number, and every number as binary. Everything is 1s and 0s!

## 2️⃣ Counting words with a dictionary (a baby language model! 👶🤖)

Count how often each word appears. This **tiny idea** is the seed of how AI learns language (we'll come back in Week 6).

```python
text = "the cat sat on the mat and the dog sat on the log"

counts = {}
for word in text.split():
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1

print(counts)
```

```python
# Which word is the most common?
most_common = ""
highest = 0
for word, number in counts.items():
    if number > highest:
        highest = number
        most_common = word
print(f"Most common: '{most_common}' ({highest} times)")
```

### `try / except` — catching the crash

**Analogy:** A safety net under a trapeze artist. 🎪 If the trick fails, they land safely instead of crashing.

```python
def safe_divide(a, b):
    try:
        return f"{a} ÷ {b} = {a / b}"
    except ZeroDivisionError:
        return "🚫 You can't divide by zero!"

print(safe_divide(10, 2))
print(safe_divide(10, 0))
```

```python
def to_number(text):
    try:
        return int(text)
    except ValueError:
        return None   # None means "nothing"

print(to_number("42"))
print(to_number("forty-two"))
```

## 4️⃣ Validation — the bouncer at the door 🚪

**Validation** checks data is **sensible** *before* using it. (It cannot prove data is *true*. An age of `"99"` passes the range check but could still be a lie.)

| Check | Question | Example |
| ----- | -------- | ------- |
| **Presence** | Is it empty? | name required |
| **Type** | Right kind of data? | age must be a number |
| **Range** | Within limits? | 1 ≤ age ≤ 120 |
| **Length** | Right size? | password ≥ 8 characters |
| **Format** | Right pattern? | email contains `@` |

```python
def register(name, age, email):
    if name.strip() == "":
        return "❌ Name is required"
    if age is None or not (1 <= age <= 120):
        return "❌ Age must be between 1 and 120"
    if "@" not in email or "." not in email:
        return "❌ That doesn't look like an email"
    return f"✅ Welcome, {name.strip().title()}!"

demo = gr.Interface(
    fn=register,
    inputs=[gr.Textbox(label="Name"), gr.Number(label="Age"), gr.Textbox(label="Email")],
    outputs="text",
    title="📝 Sign-up Form (with validation)",
    flagging_mode="never",
)
show(demo)
```

## 5️⃣ Files — saving data so it lasts

Variables **vanish** when the program ends. Files **stay**.

| Mode | Meaning |
| ---- | ------- |
| `"r"` | read |
| `"w"` | write (⚠️ **erases** what was there) |
| `"a"` | append (add to the end) |

`with open(...) as f:` closes the file automatically, like a door that shuts itself.

```python
with open("notes.txt", "w") as f:
    f.write("Learn Python\n")
    f.write("Build an AI app\n")

with open("notes.txt", "r") as f:
    for line in f:
        print("•", line.strip())
```

```python
# Notes saver: notes now survive restarting the app!
def add_note(note):
    if note.strip() != "":
        with open("notes.txt", "a") as f:
            f.write(note.strip() + "\n")
    try:
        with open("notes.txt", "r") as f:
            return f.read()
    except FileNotFoundError:
        return "(no notes yet)"

demo = gr.Interface(
    fn=add_note,
    inputs=gr.Textbox(label="New note"),
    outputs=gr.Textbox(label="All my notes", lines=8),
    title="🗒️ Notes Saver",
    flagging_mode="never",
)
show(demo)
```
